```python
import pandas as pd
import matplotlib.pyplot as plt
import numpy as np
import warnings
warnings.filterwarnings('ignore')

# Set up Vietnamese font for matplotlib
plt.rcParams['font.sans-serif'] = ['DejaVu Sans']
plt.rcParams['axes.unicode_minus'] = False

# Read the CSV file
file_path = 'Usability Test theo Task - Trang tính1.csv'
df = pd.read_csv(file_path, index_col=0)
print("Data structure:")
print(df.head(20))
```

```python
# Parse the data to extract Task information
# Task 1
task1_time_lan1 = [21, 22, 24, 25, 24, 23, 22, 18]  # Lần 1
task1_time_lan2 = [16, 30, 17, 20, 45, 17, 29, 21]  # Lần 2
task1_errors_lan1 = [0, 0, 0, 0, 0, 0, 0, 0]
task1_errors_lan2 = [0, 1, 0, 0, 2, 0, 1, 0]
task1_status_lan1 = ['Thành công', 'Thành công', 'Thành công', 'Thành công', 'Thành công', 'Thành công', 'Thành công', 'Thành công']
task1_status_lan2 = ['Thành công', 'Thành công', 'Thành công', 'Thành công', 'Lỗi', 'Thành công', 'Thành công', 'Thành công']

# Task 2
task2_time_lan1 = [55, 35, 22, 18, 37, 40, 60, 25]  # Lần 1
task2_time_lan2 = [60, 20, 33, 22, 44, 47, 67]  # Lần 2 (U8 missing)
task2_errors_lan1 = [3, 2, 1, 0, 2, 2, 3, 1]
task2_errors_lan2 = [4, 0, 2, 0, 3, 3, 3]  # U8 missing
task2_status_lan1 = ['Lỗi', 'Thành công', 'Thành công', 'Thành công', 'Thành công', 'Thành công', 'Lỗi', 'Thành công']
task2_status_lan2 = ['Lỗi', 'Thành công', 'Thành công', 'Thành công', 'Lỗi', 'Lỗi', 'Lỗi']  # U8 missing

# Task 3 (only one round)
task3_time = [20, 21, 16, 17, 14, 17, 18, 16]
task3_errors = [0, 0, 0, 0, 0, 0, 0, 0]
task3_status = ['Thành công', 'Thành công', 'Thành công', 'Thành công', 'Thành công', 'Thành công', 'Thành công', 'Thành công']

# Calculate averages
print("\n=== Average Time (seconds) ===")
print(f"Task 1 - Lần 1: {np.mean(task1_time_lan1):.2f}s")
print(f"Task 1 - Lần 2: {np.mean(task1_time_lan2):.2f}s")
print(f"Task 2 - Lần 1: {np.mean(task2_time_lan1):.2f}s")
print(f"Task 2 - Lần 2: {np.mean(task2_time_lan2):.2f}s")
print(f"Task 3: {np.mean(task3_time):.2f}s")

print("\n=== Average Errors ===")
print(f"Task 1 - Lần 1: {np.mean(task1_errors_lan1):.2f}")
print(f"Task 1 - Lần 2: {np.mean(task1_errors_lan2):.2f}")
print(f"Task 2 - Lần 1: {np.mean(task2_errors_lan1):.2f}")
print(f"Task 2 - Lần 2: {np.mean(task2_errors_lan2):.2f}")
print(f"Task 3: {np.mean(task3_errors):.2f}")
```

```python
# Chart 1: Column chart - Average execution time for both times
fig, ax = plt.subplots(figsize=(12, 6))

tasks = ['Task 1', 'Task 2', 'Task 3']
time_lan1 = [
    np.mean(task1_time_lan1),
    np.mean(task2_time_lan1),
    np.mean(task3_time)
]
time_lan2 = [
    np.mean(task1_time_lan2),
    np.mean(task2_time_lan2),
    None  # Task 3 doesn't have Lan 2 data
]

x = np.arange(len(tasks))
width = 0.35

bars1 = ax.bar(x - width/2, time_lan1, width, label='Lần 1', color='#5DADE2', edgecolor='black', linewidth=1.2)

# Only plot Lan 2 for Task 1 and 2
bars2 = ax.bar([x[0] + width/2, x[1] + width/2], [time_lan2[0], time_lan2[1]], width, 
                label='Lần 2', color='#F5B041', edgecolor='black', linewidth=1.2)

ax.set_xlabel('Task', fontsize=12, fontweight='bold')
ax.set_ylabel('Trung bình thời gian (giây)', fontsize=12, fontweight='bold')
ax.set_title('Trung bình Thời gian Thực hiện của mỗi Task', fontsize=14, fontweight='bold')
ax.set_xticks(x)
ax.set_xticklabels(tasks)
ax.legend(fontsize=11)
ax.grid(axis='y', alpha=0.3, linestyle='--')

# Add value labels on bars
for bar in bars1:
    height = bar.get_height()
    ax.text(bar.get_x() + bar.get_width()/2., height,
            f'{height:.1f}', ha='center', va='bottom', fontsize=10, fontweight='bold')

for bar in bars2:
    height = bar.get_height()
    ax.text(bar.get_x() + bar.get_width()/2., height,
            f'{height:.1f}', ha='center', va='bottom', fontsize=10, fontweight='bold')

plt.tight_layout()
plt.savefig('chart_avg_time.png', dpi=300, bbox_inches='tight')
plt.show()
print("Chart 1 saved: chart_avg_time.png")
```

```python
# Chart 2: Column chart - Average number of errors per Task
fig, ax = plt.subplots(figsize=(12, 6))

tasks = ['Task 1', 'Task 2', 'Task 3']
errors_lan1 = [
    np.mean(task1_errors_lan1),
    np.mean(task2_errors_lan1),
    np.mean(task3_errors)
]
errors_lan2 = [
    np.mean(task1_errors_lan2),
    np.mean(task2_errors_lan2),
    None  # Task 3 doesn't have Lan 2 data
]

x = np.arange(len(tasks))
width = 0.35

bars1 = ax.bar(x - width/2, errors_lan1, width, label='Lần 1', color='#52BE80', edgecolor='black', linewidth=1.2)
bars2 = ax.bar([x[0] + width/2, x[1] + width/2], [errors_lan2[0], errors_lan2[1]], width, 
                label='Lần 2', color='#EC7063', edgecolor='black', linewidth=1.2)

ax.set_xlabel('Task', fontsize=12, fontweight='bold')
ax.set_ylabel('Trung bình số lỗi', fontsize=12, fontweight='bold')
ax.set_title('Trung bình Số Lỗi của mỗi Task', fontsize=14, fontweight='bold')
ax.set_xticks(x)
ax.set_xticklabels(tasks)
ax.legend(fontsize=11)
ax.grid(axis='y', alpha=0.3, linestyle='--')

# Add value labels on bars
for bar in bars1:
    height = bar.get_height()
    ax.text(bar.get_x() + bar.get_width()/2., height,
            f'{height:.2f}', ha='center', va='bottom', fontsize=10, fontweight='bold')

for bar in bars2:
    height = bar.get_height()
    ax.text(bar.get_x() + bar.get_width()/2., height,
            f'{height:.2f}', ha='center', va='bottom', fontsize=10, fontweight='bold')

plt.tight_layout()
plt.savefig('chart_avg_errors.png', dpi=300, bbox_inches='tight')
plt.show()
print("Chart 2 saved: chart_avg_errors.png")
```

```python
# Chart 3: Stacked bar chart - Status distribution for each task
fig, ax = plt.subplots(figsize=(12, 6))

# Calculate success and failure counts for each task in both rounds
tasks_full = ['Task 1 - Lần 1', 'Task 1 - Lần 2', 'Task 2 - Lần 1', 'Task 2 - Lần 2', 'Task 3']

# Count successes and failures
success_counts = [
    sum(1 for s in task1_status_lan1 if s == 'Thành công'),
    sum(1 for s in task1_status_lan2 if s == 'Thành công'),
    sum(1 for s in task2_status_lan1 if s == 'Thành công'),
    sum(1 for s in task2_status_lan2 if s == 'Thành công'),
    sum(1 for s in task3_status if s == 'Thành công')
]

failure_counts = [
    sum(1 for s in task1_status_lan1 if s == 'Lỗi'),
    sum(1 for s in task1_status_lan2 if s == 'Lỗi'),
    sum(1 for s in task2_status_lan1 if s == 'Lỗi'),
    sum(1 for s in task2_status_lan2 if s == 'Lỗi'),
    sum(1 for s in task3_status if s == 'Lỗi')
]

x = np.arange(len(tasks_full))
width = 0.6

bars1 = ax.bar(x, success_counts, width, label='Thành công', color='#27AE60', edgecolor='black', linewidth=1.2)
bars2 = ax.bar(x, failure_counts, width, bottom=success_counts, label='Lỗi', color='#E74C3C', edgecolor='black', linewidth=1.2)

ax.set_xlabel('Task', fontsize=12, fontweight='bold')
ax.set_ylabel('Số lượng (users)', fontsize=12, fontweight='bold')
ax.set_title('Trạng thái Thực hiện của mỗi Task (Stacked Bar Chart)', fontsize=14, fontweight='bold')
ax.set_xticks(x)
ax.set_xticklabels(tasks_full, rotation=45, ha='right')
ax.legend(fontsize=11)
ax.grid(axis='y', alpha=0.3, linestyle='--')

# Add value labels on bars
for i, bar in enumerate(bars1):
    height = bar.get_height()
    if height > 0:
        ax.text(bar.get_x() + bar.get_width()/2., height/2,
                f'{int(height)}', ha='center', va='center', fontsize=10, fontweight='bold', color='white')

for i, bar in enumerate(bars2):
    height = bar.get_height()
    if height > 0:
        ax.text(bar.get_x() + bar.get_width()/2., success_counts[i] + height/2,
                f'{int(height)}', ha='center', va='center', fontsize=10, fontweight='bold', color='white')

plt.tight_layout()
plt.savefig('chart_status_distribution.png', dpi=300, bbox_inches='tight')
plt.show()
print("Chart 3 saved: chart_status_distribution.png")
```

```python
# Summary statistics
print("\n" + "="*60)
print("USABILITY TEST SUMMARY")
print("="*60)

print("\n1. AVERAGE EXECUTION TIME (seconds)")
print("-" * 60)
print(f"Task 1 - Lần 1: {np.mean(task1_time_lan1):.2f}s")
print(f"Task 1 - Lần 2: {np.mean(task1_time_lan2):.2f}s")
print(f"Task 1 - Improvement: {((np.mean(task1_time_lan1) - np.mean(task1_time_lan2)) / np.mean(task1_time_lan1) * 100):.1f}%")
print()
print(f"Task 2 - Lần 1: {np.mean(task2_time_lan1):.2f}s")
print(f"Task 2 - Lần 2: {np.mean(task2_time_lan2):.2f}s")
print(f"Task 2 - Difference: {((np.mean(task2_time_lan2) - np.mean(task2_time_lan1)) / np.mean(task2_time_lan1) * 100):.1f}%")
print()
print(f"Task 3: {np.mean(task3_time):.2f}s")

print("\n2. AVERAGE NUMBER OF ERRORS")
print("-" * 60)
print(f"Task 1 - Lần 1: {np.mean(task1_errors_lan1):.2f} errors")
print(f"Task 1 - Lần 2: {np.mean(task1_errors_lan2):.2f} errors")
print()
print(f"Task 2 - Lần 1: {np.mean(task2_errors_lan1):.2f} errors")
print(f"Task 2 - Lần 2: {np.mean(task2_errors_lan2):.2f} errors")
print()
print(f"Task 3: {np.mean(task3_errors):.2f} errors")

print("\n3. SUCCESS RATE BY TASK")
print("-" * 60)
print(f"Task 1 - Lần 1: {(sum(1 for s in task1_status_lan1 if s == 'Thành công') / len(task1_status_lan1) * 100):.1f}% ({sum(1 for s in task1_status_lan1 if s == 'Thành công')}/{len(task1_status_lan1)})")
print(f"Task 1 - Lần 2: {(sum(1 for s in task1_status_lan2 if s == 'Thành công') / len(task1_status_lan2) * 100):.1f}% ({sum(1 for s in task1_status_lan2 if s == 'Thành công')}/{len(task1_status_lan2)})")
print()
print(f"Task 2 - Lần 1: {(sum(1 for s in task2_status_lan1 if s == 'Thành công') / len(task2_status_lan1) * 100):.1f}% ({sum(1 for s in task2_status_lan1 if s == 'Thành công')}/{len(task2_status_lan1)})")
print(f"Task 2 - Lần 2: {(sum(1 for s in task2_status_lan2 if s == 'Thành công') / len(task2_status_lan2) * 100):.1f}% ({sum(1 for s in task2_status_lan2 if s == 'Thành công')}/{len(task2_status_lan2)})")
print()
print(f"Task 3: {(sum(1 for s in task3_status if s == 'Thành công') / len(task3_status) * 100):.1f}% ({sum(1 for s in task3_status if s == 'Thành công')}/{len(task3_status)})")

print("\n" + "="*60)
```
